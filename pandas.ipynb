Filtering Data

```python
data = {
'id': [1,2,3,4],
'name':["asmath","aisha","adina","hani"],
'marks':[72,80,49,58],
'ages':[22,19,20,21],
'department':["HR","IT","HR","IT"]
}
df = pd.DataFrame(data,index=['a','b','c','d'])
df
```

```python
df['marks']>50
```

```python
df[df['marks']>50]
```

```python
df[df['marks']>50][['name','department']]
```

```python
df.loc[df['marks']>50,['name','department']]
```

```python
df.loc[df['name'].str.startswith('a'),['department','name']]
```

```python
df.loc[df['name'].str.len()>4]
```

```python
df.loc[(df['marks']>50) & (df['department']=='HR')]
```

Select all students who have marks greater than 50 and their department is

###in,notin
* and
* isin

```python
df.loc[(df['marks']>50) & (df['department'].isin(['IT','HR']))]
```

Not in

```python
df.loc[(df['marks']>50) & ~(df['department'].isin(['HR']))]
```

```python
df[df['marks'].between(55,73)]
```

Apply Function

```python
def gradeDef(marks):
    if marks>80:
        return'A'
    elif marks>60:
        return'B'
    elif marks>40:
        return'C'
    else:
        return'F'
print(gradeDef(40))    
```

```python
df['Grade']=df['marks'].apply(gradeDef)
df
```

```python
df['Bonus']=df['marks'].apply(lambda x:x+5)
df
```

Join in Python (Merge)

```python
students = pd.DataFrame({'student_id':[1,2,3],'name':['Asmath','Aisha','Adina']})
marks = pd.DataFrame({'student_id':[1,2,3,4],'marks':[100,50,20,90]})
```

```python
pd.merge(
    students,
    marks,
    on='student_id',
    how='inner'
)
```

```python
pd.merge(
    students,
    marks,
    on='student_id',
    how='outer'
)
```
