Aggregate Function


```python
d = np.array([10,20,30,40])
print(np.sum(d))
print(d.sum())
print(np.max(d))
print(np.min(d))
print(np.mean(d))
print(np.median(d))
e = np.array([10,30,20,40,50])
print(np.median(e))

```

Working with 2d arrays


axis = 0  column

axis = 1  row

axis = none(all over)

```python
arr_2d = np.array([[1,2],[2,3],[4,5]])
print(np.sum(arr_2d,axis = 0))
print(np.sum(arr_2d,axis = 1))
print(np.sum(arr_2d,axis = None))
print(np.mean(arr_2d,axis = 1))
```

Filtering

```python
x=np.array([1,2,10,20])
print((x>=10) & (x<20))
print((x>=10) | (x<20))
```

```python
y = np.array([20,10,30,5])
x = np.where(y==10,'GOOD','BAD')
print(x)
```
